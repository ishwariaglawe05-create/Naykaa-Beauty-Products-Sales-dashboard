```python
# Online Shopping Product Management System
# Using Class and Object

class Product:
    def __init__(self, product_id, name, price, quantity):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.quantity = quantity

    def display_product(self):
        print("Product ID:", self.product_id)
        print("Product Name:", self.name)
        print("Price:", self.price)
        print("Quantity:", self.quantity)
        print("------------------------")


class Shopping:
    def __init__(self):
        self.products = []

    def add_product(self):
        product_id = len(self.products) + 1

        name = input("Enter product name: ")
        price = float(input("Enter product price: "))
        quantity = int(input("Enter quantity: "))

        product = Product(product_id, name, price, quantity)

        self.products.append(product)

        print("Product added successfully!")

    def display_products(self):
        if len(self.products) == 0:
            print("No products available.")
        else:
            print("\n===== PRODUCT LIST =====")

            for product in self.products:
                product.display_product()

    def search_product(self):
        name = input("Enter product name to search: ")

        for product in self.products:

            if product.name.lower() == name.lower():
                print("\nProduct Found!")
                product.display_product()
                return

        print("Product not found.")

    def buy_product(self):
        product_id = int(input("Enter product ID: "))
        buy_quantity = int(input("Enter quantity: "))

        for product in self.products:

            if product.product_id == product_id:

                if buy_quantity <= product.quantity:

                    total = product.price * buy_quantity
                    product.quantity = product.quantity - buy_quantity

                    print("\n===== BILL =====")
                    print("Product:", product.name)
                    print("Price:", product.price)
                    print("Quantity:", buy_quantity)
                    print("Total Amount:", total)
                    print("Purchase successful!")

                else:
                    print("Not enough stock.")

                return

        print("Product ID not found.")


# Create Object of Shopping Class
shopping = Shopping()


# Main Menu
while True:

    print("\n===== ONLINE SHOPPING SYSTEM =====")
    print("1. Add Product")
    print("2. Display Products")
    print("3. Search Product")
    print("4. Buy Product")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        shopping.add_product()

    elif choice == "2":
        shopping.display_products()

    elif choice == "3":
        shopping.search_product()

    elif choice == "4":
        shopping.buy_product()

    elif choice == "5":
        print("Thank you for using Online Shopping System!")
        break

    else:
        print("Invalid choice. Please try again.")
```
